```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/kaggle/input/iris-flower-dataset/IRIS.csv')
```

```python
df.head()
```

```python
df.isna().sum()
```

all have not null values,lessfeature so all are important ,all in the numric form so no need to preprocessing

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('species',axis = 1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.33, random_state=42)
```

```python

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

knn and svm model are distance based so we need to standerdlize to avoid outlier

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=13)
```

```python
model_knn.fit(X_train,y_train)
```

```python
model_knn.score(X_test,y_test)
```

```python
from sklearn.svm import SVC
```

```python
model_svm = SVC(gamma= 'auto')
```

```python

model_svm.fit(X_train,y_train)
```

```python
model_svm.score(X_test,y_test)
```

uses of grid searchcv

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model_rf = RandomForestClassifier()
```

```python
model_rf.fit(X_train,y_train)
```

```python
model_rf.score(X_test,y_test)
```

all model gives the 98% accuracy this is also called Overfitting . for avoid this condition we use cross validation

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model_rf, X, y, cv=5)
print(scores.mean())
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model_svm, X, y, cv=5)
print(scores.mean())
```

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(model_knn, X, y, cv=5)
print(scores.mean())
```

from cross validation we conclude that svm can give best result
